```python
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. Chemins (On reste cohérent avec l'arborescence)
BASE_DIR = ".."
RAW_DATA_DIR = os.path.join(BASE_DIR, "data", "raw", "plantvillage dataset", "color")

# 2. Paramètres techniques (Standards pour la Vision par Ordinateur)
IMG_SIZE = (224, 224) # Taille requise par la plupart des modèles d'IA (ex: MobileNet, ResNet)
BATCH_SIZE = 32       # On traite les images 32 par 32 pour ne pas saturer la RAM

# 3. Préparation du "Data Generator"
# On inclut la normalisation (rescale) : transforme les pixels [0-255] en [0-1]
# On réserve 20% des images pour la validation (test de performance)
datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

print("⏳ Chargement du flux de données...")

# 4. Chargement des données d'entraînement
train_generator = datagen.flow_from_directory(
    RAW_DATA_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    shuffle=True
)

# 5. Chargement des données de validation (pour tester l'IA plus tard)
validation_generator = datagen.flow_from_directory(
    RAW_DATA_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    shuffle=False
)

# 6. Vérification finale
print(f"\n✅ Préparation terminée.")
print(f"Nombre de classes détectées : {train_generator.num_classes}")
```
