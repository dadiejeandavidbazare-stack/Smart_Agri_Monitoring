```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os

# 1. CONFIGURATION
MODEL_PATH = '../models/agri_model_v1.h5'
DATA_DIR = "../data/raw/plantvillage dataset/color"

# 2. CHARGEMENT (Modèle + Liste des noms de maladies)
print("⏳ Chargement de l'intelligence artificielle...")
model = tf.keras.models.load_model(MODEL_PATH)
class_names = sorted(os.listdir(DATA_DIR))
print(f"✅ Prêt ! {len(class_names)} maladies répertoriées.")

# 3. FONCTION DE DIAGNOSTIC
def run_diagnostic(img_path):
    # Prétraitement de l'image (Format MobileNetV2)
    img = tf.keras.preprocessing.image.load_img(img_path, target_size=(224, 224))
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) / 255.0

    # Prédiction
    preds = model.predict(img_array)
    idx = np.argmax(preds[0])
    confidence = np.max(preds[0]) * 100
    label = class_names[idx]

    # Affichage
    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.title(f"RÉSULTAT : {label}\nCONFIANCE : {confidence:.2f}%", color='green' if confidence > 80 else 'orange')
    plt.axis('off')
    plt.show()
    
    return label, confidence

# 4. TEST SUR UNE IMAGE (Modifie l'index [10] pour changer de plante)
test_folder = class_names[10] 
sample_img = os.listdir(os.path.join(DATA_DIR, test_folder))[0]
path_to_test = os.path.join(DATA_DIR, test_folder, sample_img)

print(f"🔍 Analyse de l'image : {sample_img}")
run_diagnostic(path_to_test)
```
